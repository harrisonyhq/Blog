# 🟢 Easy: Implement ReLU

Implement the **ReLU** (Rectified Linear Unit) activation function from scratch.

$$\text{ReLU}(x) = \max(0, x)$$

### Signature
```python
def relu(x: torch.Tensor) -> torch.Tensor:
    ...
```

### Rules
- Do **NOT** use `torch.relu`, `F.relu`, `torch.clamp`, or any built-in activation
- Must support autograd (gradients should flow back)

### Example
```
Input:  tensor([-2., -1., 0., 1., 2.])
Output: tensor([ 0.,  0., 0., 1., 2.])
```

````python
import torch
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def relu(x: torch.Tensor) -> torch.Tensor:
    return x * (x > 0)
````

````python
# 🧪 Test your implementation (feel free to add more debug prints)
x = torch.tensor([-2., -1., 0., 1., 2.])
print("Input: ", x)
print("Output:", relu(x))
print("Shape: ", relu(x).shape)
````

````python
x1 = torch.tensor(0.,requires_grad=True)
y = torch.relu(x1)
y.backward()
print(x1.grad)
x2 = torch.tensor(0.,requires_grad=True)
z = relu(x2)
z.backward()
print(x2.grad)
def relu_max(x: torch.Tensor) -> torch.Tensor:
    return torch.max(x, torch.zeros_like(x))
x3 = torch.tensor(0.,requires_grad=True)
m = relu_max(x3)
m.backward()
print(x3.grad)

'''
上述代码运行结果：
tensor(0.)
tensor(0.)
tensor(0.5000)
在深度学习中，反向传播过程中主要计算loss对每个参数的梯度，优化器根据梯度来更新参数。
标准 torch.relu 通常在 x = 0 处取梯度 0。但是如果采用torch.max(x, torch.zeros_like(x))实现，
前向结果虽然一样，但在 x = 0 时，max(0, 0) 两边相等，PyTorch 会把梯度平均分给两个输入，
所以传回 x 的梯度是 0.5。这会导致它和标准 ReLU 的反向行为不完全一致。
'''
````
